```python
import re

html_string = "<h1>Welcome to Python Programming</h1>"
# re.sub replaces all occurrences matching the pattern with an empty string.

def remove_html(text):
    clean_text = re.sub(r"<.*?>", "", text)
    return clean_text

remove_html(html_string)
```

```python
all_docs = list()

for page in range(1, 81):
    with open(f"india-gk\\page-{page}.html", "r", encoding='utf-8')  as f:
        html_content = f.read()
        
    soup  = BeautifulSoup(html_content, 'html.parser')
    
    questions = soup.find_all("b")
    list_of_ans = soup.find_all("ul", class_="list-unstyled")
    right_option = soup.find_all("button", class_="btn")
    

    for idx in range(len(questions)):
        new = dict()
        new["question"] = remove_html(str(questions[idx]))
        new["answers"] = remove_html(str(list_of_ans[idx]))
        new["right_ans"] = str(right_option[idx]).split("data-answer")[-1][2:3]
        all_docs.append(new)


import json

with open("answers.json", "w", encoding='utf-8') as f :
    json.dump(all_docs, f, ensure_ascii=False)
```
